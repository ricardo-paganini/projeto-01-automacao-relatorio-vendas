## 9. Ingestão dos dados — camada Bronze

### 9.1 Objetivo

Carregar as planilhas VENDAS, PRODUTOS e VENDEDORES do arquivo Excel armazenado no Volume do Unity Catalog e persistir os dados em tabelas Delta.

### 9.2 Estratégia

A ingestão deverá preservar os registros de origem, sem aplicar regras de limpeza ou remover duplicidades. Serão adicionados metadados técnicos para identificar o arquivo de origem e o momento da ingestão.

### 9.3 Tabelas de destino

- `workspace.default.bronze_vendas`
- `workspace.default.bronze_produtos`
- `workspace.default.bronze_vendedores`

### 9.4 Critérios de validação

- As três tabelas são criadas com sucesso.
- As contagens gravadas são comparadas com as contagens lidas.
- Os cabeçalhos e os tipos de dados são verificados.
- Nenhum tratamento de negócio é aplicado nesta camada.

```python
caminho_arquivo = (
    '/Volumes/workspace/default/projeto_vendas/dados_projeto_01.xlsx'
)

nomes_planilhas = ['vendas', 'produtos', 'vendedores']
dados_origem = {}

workbook = load_workbook(
    filename=caminho_arquivo,
    read_only=True,
    data_only=True
)

try:
    for nome in nomes_planilhas:
        planilha = workbook[nome]
        linhas = planilha.iter_rows(values_only=True)

        cabecalhos = list(next(linhas))
        registros = [
            tuple(linha)
            for linha in linhas
            if any(valor is not None for valor in linha)
        ]

        dados_origem[nome] = {
            'cabecalhos': cabecalhos,
            'registros': registros
        }

        print(
            f'{nome}: {len(registros)} registros; '
            f'{len(cabecalhos)} colunas'
        )
finally:
    workbook.close()
```

### 9.7 Diagnóstico de incompatibilidade de tipos

#### Problema identificado

Durante a criação dos DataFrames Spark a partir dos registros do Excel, ocorreu o erro `CANNOT_MERGE_TYPE`.

A mensagem indica que o Spark encontrou valores que exigem tipos numéricos diferentes (`DoubleType` e `LongType`) ao tentar inferir automaticamente o esquema de uma ou mais colunas.

#### Impacto

A conversão dos registros em DataFrames não foi concluída. Portanto, os DataFrames Bronze ainda não foram validados e nenhuma tabela Bronze foi gravada nesta atividade.

#### Investigação e decisão técnica

Será realizada uma inspeção dos tipos presentes nas colunas das três fontes. Com base nos resultados, será definido um esquema explícito para cada DataFrame, respeitando o significado dos campos e preservando os valores de origem.

Nenhum registro será excluído ou convertido arbitrariamente para contornar o erro.

#### Status

**Pendente:** identificar as colunas afetadas, definir os esquemas e validar a criação dos DataFrames Spark.

```python
for nome, dados in dados_origem.items():
    print(f'\nFonte: {nome}')

    for indice, coluna in enumerate(dados['cabecalhos']):
        tipos = Counter(
            type(registro[indice]).__name__
            for registro in dados['registros']
            if registro[indice] is not None
        )

        if len(tipos) > 1:
            print(f'\nColuna: {coluna}')
            print(f'Tipos encontrados: {dict(tipos)}')

            exemplos = [
                (registro[indice], type(registro[indice]).__name__)
                for registro in dados['registros']
                if registro[indice] is not None
            ]

            print('Exemplos:', exemplos[:10])
```

#### Resultado da investigação

A inspeção dos valores identificou tipos Python mistos em duas colunas da fonte `VENDAS`:

- `preco_unitario`: 9.906 valores identificados como `float` e 106 como `int`.
- `desconto`: 6.990 valores identificados como `float` e 3.007 como `int`.

As fontes `PRODUTOS` e `VENDEDORES` não apresentaram colunas com mais de um tipo Python entre os valores não nulos nesta inspeção.

A variação entre inteiros e decimais é compatível com valores como preços e descontos representados sem casas decimais quando não são necessárias. O problema ocorreu na inferência automática do esquema pelo Spark, e não constitui, por si só, evidência de dados inválidos.

**Decisão técnica proposta:** definir um esquema explícito para cada DataFrame, usando tipos adequados ao significado dos campos. Para valores monetários e percentuais, avaliar a utilização de `DecimalType` para controlar a precisão, evitando conversões arbitrárias.

### 9.8 Inspeção dos tipos e valores de origem

#### Objetivo

Investigar os tipos de dados presentes nas colunas relevantes antes de definir os esquemas dos DataFrames Spark.

#### Justificativa técnica

A inferência automática do Spark falhou porque as colunas `preco_unitario` e `desconto` contêm valores inteiros e decimais. Antes de estabelecer os tipos explicitamente, é necessário verificar também os campos de identificação, quantidade, data, custo e preço-base.

#### Procedimento

A inspeção verifica:

- Os tipos Python encontrados nos valores não nulos.
- A quantidade de valores nulos por coluna.
- Exemplos dos valores existentes, para identificar possíveis inconsistências.

Essa atividade é exclusivamente diagnóstica. Os valores originais não serão modificados, convertidos ou descartados.

#### Decisão técnica

Os esquemas serão definidos após a análise dos resultados, considerando o significado de cada campo e a necessidade de preservar a precisão dos valores monetários e percentuais.

#### Critério de validação

A inspeção deverá apresentar os tipos encontrados, a contagem de valores nulos e exemplos das colunas relevantes. A definição dos esquemas permanecerá pendente até que esses resultados sejam avaliados.

```python
for nome, dados in dados_origem.items():
    print(f'\nFonte: {nome}')

    for indice, coluna in enumerate(dados['cabecalhos']):
        valores = [
            registro[indice]
            for registro in dados['registros']
            if registro[indice] is not None
        ]

        tipos = Counter(type(valor).__name__ for valor in valores)

        if coluna in {
            'pedido_id', 'data_venda', 'cliente_id',
            'produto_id', 'vendedor_id', 'quantidade',
            'preco_unitario', 'desconto', 'custo', 'preco_base'
        }:
            print(f'\nColuna: {coluna}')
            print(f'Tipos: {dict(tipos)}')
            print(f'Valores nulos: {len(dados["registros"]) - len(valores)}')
            print(f'Exemplos: {valores[:5]}')
```

### 9.10 Diagnóstico da criação dos DataFrames Spark

#### Objetivo

Investigar a falha `AssertionError` ocorrida durante a criação do DataFrame da fonte `VENDAS`, utilizando esquemas explícitos.

#### Contexto técnico

O ambiente Databricks utiliza Spark Connect no compute Serverless. Durante a tentativa de criação dos DataFrames, a fonte `VENDAS` apresentou um `AssertionError`. O aviso relacionado ao parâmetro `verifySchema` também indicou que esse parâmetro não é suportado pelo Spark Connect.

A causa exata da falha ainda não foi identificada.

A investigação será dividida em três verificações:

1. Comparar a quantidade e a ordem das colunas da origem com o esquema definido.
2. Inspecionar os tipos Python presentes nos registros e compará-los aos tipos Spark esperados.
3. Repetir a criação do DataFrame com uma amostra pequena para verificar se a falha ocorre mesmo com poucos registros.

#### 9.11 Causa identificada e correção planejada

O traceback revelou que a falha ocorre na conversão dos valores para `DecimalType`. A função interna de conversão do Spark Connect exige valores do tipo Python `decimal.Decimal`, mas os registros de origem contêm valores `float` e `int` nessas colunas.

**Causa identificada:** incompatibilidade entre os tipos dos valores Python e o tipo decimal definido no esquema Spark.

**Correção planejada:** converter explicitamente os valores monetários e percentuais para `decimal.Decimal`, preservando os valores ausentes. A conversão será realizada antes da criação dos DataFrames.

**Critério de validação:** os três DataFrames deverão ser criados com sucesso, mantendo as contagens esperadas e os tipos definidos nos esquemas.

**Status:** causa identificada; correção pendente de execução e validação.

```python
colunas_decimais = {
    'vendas': ['preco_unitario', 'desconto'],
    'produtos': ['custo', 'preco_base'],
    'vendedores': [],
}

dados_preparados = {}

for nome, dados in dados_origem.items():
    indices_decimais = {
        indice
        for indice, coluna in enumerate(dados['cabecalhos'])
        if coluna in colunas_decimais[nome]
    }

    registros = []

    for registro in dados['registros']:
        valores = list(registro)

        for indice in indices_decimais:
            valor = valores[indice]

            if valor is not None:
                valores[indice] = Decimal(str(valor))

        registros.append(tuple(valores))

    dados_preparados[nome] = registros

print('Conversão decimal concluída para as fontes preparadas.')
```

#### 9.12 Preparação dos valores decimais

Os valores das colunas monetárias e percentuais foram convertidos explicitamente para `decimal.Decimal` antes da criação dos DataFrames Spark.

Essa conversão atende ao requisito de tipo do Spark Connect para colunas definidas como `DecimalType`, evitando a falha identificada na etapa de diagnóstico.

**Resultado observado:** a célula de preparação foi executada sem erros.

**Validação pendente:** a criação dos DataFrames, a conferência das contagens e a inspeção dos esquemas ainda precisam ser concluídas.

```python

dataframes_bronze = {}

for nome, dados in dados_origem.items():
    df = spark.createDataFrame(
        dados_preparados[nome],
        schema=esquemas[nome]
    )

    quantidade = df.count()
    dataframes_bronze[nome] = df

    print(f'\nFonte: {nome}')
    print(f'Registros: {quantidade}')
    df.printSchema()
```
