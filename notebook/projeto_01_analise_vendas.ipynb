# Projeto 01 — Retail Sales Analytics Platform

## 1. Contextualização do negócio

### 1.1 Cenário

Este projeto simula uma empresa varejista que comercializa produtos por diferentes canais de venda e possui vendedores distribuídos por regiões.

As informações comerciais estão organizadas em três planilhas Excel: vendas, produtos e vendedores. O objetivo é estruturar esses dados em uma plataforma analítica que permita avaliar o desempenho comercial e apoiar a tomada de decisão.

Dados comerciais armazenados em planilhas não estão necessariamente prontos para análise. Diferenças de preenchimento, registros duplicados, valores ausentes e inconsistências podem comprometer a confiabilidade dos indicadores.

O projeto busca construir um fluxo reproduzível de ingestão, validação, tratamento e organização dos dados, com rastreabilidade das transformações realizadas.

### 1.3 Objetivo geral

Desenvolver uma solução de engenharia e análise de dados utilizando o Databricks Free, Python, PySpark e SQL, transformando os dados de origem em tabelas analíticas e dashboards para acompanhamento do desempenho comercial.

### 1.4 Perguntas de negócio

A solução deverá permitir investigar:

- Como a receita evolui ao longo do período analisado?
- Quais produtos e categorias apresentam maior desempenho?
- Quais canais de venda geram maior receita?
- Como se distribuem os resultados entre vendedores e regiões?
- Como descontos e status dos pedidos se relacionam com os indicadores comerciais?

As respostas serão apresentadas somente após a preparação e validação dos dados.

## 9. Ingestão dos dados — camada Bronze

### 9.1 Objetivo

Carregar as planilhas VENDAS, PRODUTOS e VENDEDORES do arquivo Excel armazenado no Volume do Unity Catalog e persistir os dados em tabelas Delta.

### 9.2 Estratégia

A ingestão deverá preservar os registros de origem, sem aplicar regras de limpeza ou remover duplicidades. Serão adicionados metadados técnicos para identificar o arquivo de origem e o momento da ingestão.

### 9.3 Tabelas de destino

- `workspace.default.bronze_vendas`
- `workspace.default.bronze_produtos`
- `workspace.default.bronze_vendedores`

### 9.4 Critérios de validação

- As três tabelas são criadas com sucesso.
- As contagens gravadas são comparadas com as contagens lidas.
- Os cabeçalhos e os tipos de dados são verificados.
- Nenhum tratamento de negócio é aplicado nesta camada.

```python
caminho_arquivo = (
    '/Volumes/workspace/default/projeto_vendas/dados_projeto_01.xlsx'
)

nomes_planilhas = ['vendas', 'produtos', 'vendedores']
dados_origem = {}

workbook = load_workbook(
    filename=caminho_arquivo,
    read_only=True,
    data_only=True
)

try:
    for nome in nomes_planilhas:
        planilha = workbook[nome]
        linhas = planilha.iter_rows(values_only=True)

        cabecalhos = list(next(linhas))
        registros = [
            tuple(linha)
            for linha in linhas
            if any(valor is not None for valor in linha)
        ]

        dados_origem[nome] = {
            'cabecalhos': cabecalhos,
            'registros': registros
        }

        print(
            f'{nome}: {len(registros)} registros; '
            f'{len(cabecalhos)} colunas'
        )
finally:
    workbook.close()
```

### 9.5 Resultado da leitura das fontes

A leitura do arquivo Excel foi executada no Databricks utilizando `openpyxl`. Os cabeçalhos foram separados dos registros e as linhas completamente vazias foram desconsideradas.

| Fonte | Registros lidos | Colunas |
|---|---:|---:|
| VENDAS | 10.012 | 10 |
| PRODUTOS | 20 | 6 |
| VENDEDORES | 10 | 3 |

**Validação:** as três planilhas foram lidas e as contagens de registros foram exibidas sem incluir os cabeçalhos.

**Limitação:** esta atividade valida a leitura do arquivo, mas ainda não confirma a gravação persistente em Delta Lake nem a qualidade dos dados.

**Status:** leitura das fontes concluída; gravação Bronze pendente.
