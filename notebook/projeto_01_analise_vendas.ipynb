# Projeto 01 — Retail Sales Analytics Platform

## 1. Contextualização do negócio

### 1.1 Cenário

Este projeto simula uma empresa varejista que comercializa produtos por diferentes canais de venda e possui vendedores distribuídos por regiões.

As informações comerciais estão organizadas em três planilhas Excel: vendas, produtos e vendedores. O objetivo é estruturar esses dados em uma plataforma analítica que permita avaliar o desempenho comercial e apoiar a tomada de decisão.

### 1.2 Problema de negócio

Dados comerciais armazenados em planilhas não estão necessariamente prontos para análise. Diferenças de preenchimento, registros duplicados, valores ausentes e inconsistências podem comprometer a confiabilidade dos indicadores.

O projeto busca construir um fluxo reproduzível de ingestão, validação, tratamento e organização dos dados, com rastreabilidade das transformações realizadas.

### 1.3 Objetivo geral

Desenvolver uma solução de engenharia e análise de dados utilizando o Databricks Free, Python, PySpark e SQL, transformando os dados de origem em tabelas analíticas e dashboards para acompanhamento do desempenho comercial.

### 1.4 Perguntas de negócio

A solução deverá permitir investigar:

- Como a receita evolui ao longo do período analisado?
- Quais produtos e categorias apresentam maior desempenho?
- Quais canais de venda geram maior receita?
- Como se distribuem os resultados entre vendedores e regiões?
- Como descontos e status dos pedidos se relacionam com os indicadores comerciais?

As respostas serão apresentadas somente após a preparação e validação dos dados.

## 2. Escopo do projeto

### 2.1 Incluído

- Ingestão dos dados de origem no Databricks.
- Organização dos dados nas camadas Bronze, Silver e Gold.
- Avaliação da qualidade dos dados.
- Tratamento documentado de inconsistências.
- Modelagem de tabelas analíticas com Delta Lake.
- Consultas e análises com PySpark e SQL.
- Construção de dashboards nativos do Databricks, conforme os recursos disponíveis no ambiente.
- Documentação técnica e versionamento do projeto com Git e GitHub.

### 2.2 Fora do escopo inicial

- Modelos preditivos e machine learning.
- Integrações com sistemas corporativos.
- Implantação em ambiente produtivo empresarial.
- Dependência de funcionalidades pagas do Databricks.

## 3. Fonte de dados

### 3.1 Arquivo de origem

O arquivo `dados_projeto_01.xlsx` foi disponibilizado em um Volume do Unity Catalog.

Caminho utilizado no ambiente:

`/Volumes/workspace/default/projeto_vendas/dados_projeto_01.xlsx`

### 3.2 Estrutura identificada

| Planilha | Linhas no Excel | Colunas |
|---|---:|---:|
| VENDAS | 10.013 | 10 |
| PRODUTOS | 21 | 6 |
| VENDEDORES | 11 | 3 |

As dimensões foram verificadas no Databricks. As contagens incluem a linha de cabeçalho; a quantidade efetiva de registros será confirmada durante a ingestão.

### 3.3 Descrição das fontes

**VENDAS:** contém identificadores de pedido, data, cliente, produto e vendedor, além de quantidade, preço unitário, desconto, canal de venda e status.

**PRODUTOS:** contém identificador, nome, categoria, subcategoria, custo e preço-base dos produtos.

**VENDEDORES:** contém identificador, nome e região dos vendedores.

### 3.4 Limitações iniciais

A granularidade exata dos registros, as regras de negócio e a qualidade dos dados ainda serão investigadas. Nenhum registro será removido ou alterado sem uma justificativa documentada.

## 5. Arquitetura proposta

O fluxo analítico será organizado em três camadas:

- **Bronze:** ingestão dos dados de origem, preservando os registros recebidos e adicionando metadados técnicos de ingestão.
- **Silver:** validação, padronização e tratamento documentado dos dados.
- **Gold:** tabelas analíticas preparadas para consultas de negócio e dashboards.

A arquitetura será implementada progressivamente, validando cada camada antes de avançar para a seguinte.

## 6. Metodologia

O projeto seguirá os princípios do CRISP-DM, adaptados ao escopo:

1. Entendimento do negócio.
2. Entendimento dos dados.
3. Preparação dos dados.
4. Modelagem analítica, quando necessária.
5. Avaliação dos dados e resultados.
6. Disponibilização dos indicadores e documentação.

Cada etapa terá objetivo, implementação, resultado observado, critérios de validação e registro das limitações.

## 7. Controle de execução

As etapas serão consideradas concluídas somente após a execução do código e a validação dos resultados.

Os resultados numéricos, as decisões de tratamento e os indicadores serão documentados com base nas evidências produzidas no Databricks. Etapas ainda não executadas serão identificadas como pendentes.
